```python
import re
from nltk.util import ngrams

text =( """AI is simulation of human intelliengence by machine.
It includes expects systems NLP speech recognition and machine vision""")

# Convert the sentence to lowercase and tokenize
words = re.findall(r"\b[\w-]+\b",text.lower())

# Generate bigrams
bigrams = list(ngrams(words, 2))

# Generate trigrams
trigrams = list(ngrams(words, 3))

print("Bigrams:")
for bigram in bigrams:
    print(bigram)

print("\nTrigrams:")
for trigram in trigrams:
    print(trigram)
```
